```python
import ollama


model_name = 'gpt-oss:20b'
small_model_name = 'deepseek-r1:1.5b'
# Pull the model
!ollama pull {model_name}
!ollama pull {small_model_name}
```

```python
questions = ["what is SMILES in cheminformatics? Mention several key syntax features. Keep your answer concise.",
    "if aqueous solubility was increased, how does logP change",
 "MMP transform: R=H → R=OH on a saturated alkyl carbon. Predict the direction of change for aqueous solubility and logP (increase/decrease)."]



```

```python
#https://python.langchain.com/docs/integrations/llms/ollama/


from langchain_ollama.llms import OllamaLLM
from langchain_core.messages import HumanMessage

model_gptoss = OllamaLLM(model= model_name)
model_small = OllamaLLM(model= small_model_name)

# test the LLM itself works
for q in questions:
  print('='* 40)
  print()
  print("Question: ", q)
  print()
  result = model_small.invoke([HumanMessage(content= q,name="demo")])
  print('small_model:\n',result)
  print('-'*20)
  print()
  result = model_gptoss.invoke([HumanMessage(content= q ,name="demo")])
  print('gptoss:\n',result)
  print()
  print('='* 40)

```

```python
# if you want think

full_answer = model_small.generate(['What is logP for aqueous solubility '])
generate = full_answer.generations[0][0]
print('<think>')
print(generate.generation_info.get('thinking', '') )
print('</think>')
print('<answer>')
print(generate.text)
print('</answer>')
```
